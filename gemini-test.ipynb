```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import (
    HumanMessage,
    SystemMessage,
    RemoveMessage,
)
from dotenv import load_dotenv
import os

load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.1-flash-lite-preview",
    google_api_key=os.environ["GOOGLE_API_KEY"],
)
```

```python
sys_msg_content = (
        "You are a helpful, intelligent AI assistant. "
        "Answer general questions using internal knowledge. "
        "Use tools only when strictly necessary."
    )
user_msg_content = "What is the capital of France?"
```

```python
final_messages = [SystemMessage(content=sys_msg_content)] + [HumanMessage(content=user_msg_content)]
response = llm.invoke(final_messages)
```

```python
print(response.content)
```
